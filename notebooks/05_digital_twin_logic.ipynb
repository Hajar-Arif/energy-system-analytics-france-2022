###**SECTION 5: Digital twin logic**

In this section, we will define the logic behind a digital twin for the French electricity grid: system variables, states, evolution rules, scenario simulations, and critical assumptions.

**1. What is a digital twin (Power grid) ?**

A digital twin is a virtual representation of a physical system that mirrors its behavior in real time.
For the French electricity grid, this means building a model that can simulate how the system responds to changes, such as a nuclear outage, or a winter cold wave, before they actually happen.

This notebook does not implement a working digital twin. Instead, it defines the logic behind it: what states the system can be in, what rules govern its evolution, and what scenarios we would want to simulate.

**2. System variables**

At any given moment, the French electricity system can be described by a number of variables that define the current state of the system:

- Demand level (= Total consumption): represented in our dataset by the column "Consommation".

- Production mix: represented in our dataset by the columns	"Nucléaire, Gaz, Eolien, Solaire, Hydraulique, etc.".

- System balance (= Surplus or deficit): represented in our dataset by the difference between the columns	"*Total_Production (calculated column)*" and "Consommation".

- Exchange direction (= importer or exporter):	represented in our dataset by the column "Ech. physiques".

- Carbon intensity:	represented in our dataset by the column "Taux de Co₂".

- Nuclear availability: represented in our dataset by the ratio between the columns "Nucléaire" and *"Installed capacity (estimated constant)"*.

**3. System states**

The above variables determine which state the system is in:

- Sufficient: production covers demand, no imports needed.
- Deficit: production falls short, imports required.
- Surplus: excess production, available for export.
- Crisis: multiple stress factors align (low nuclear + peak demand + high imports).

**A digital twin must detect the current state in real time and predict transitions between states, because the system is most vulnerable during the shifts.**

**4. Evolution rules**

The system follows rules, which are cause-and-effect relationships that govern how the system transitions from one state to another:

*- Rule 1: demand drives everything.*
When consumption increases (cold waves for example), the system must respond by increasing production or imports.

*Rule 2: nuclear is the backbone of the system.*
If the nuclear output drops, the system loses its baseload and must compensate with gas, imports, or both (which may increase the CO2 emissions). This has been the most common trigger for state transitions and increased imports in France during the year 2022.

*Rule 3: Renewables are volatile.*
Wind and solar can disappear within hours. The digital twin must account for this intermittency.

*Rule 4: Imports are the safety alternative. *
When domestic production cannot meet demand, physical exchange with neighboring european countries kicks in. But this depends on their availability too, if all of Europe is in a cold wave, imports may not be possible.

**These rules are the logic a digital twin must encode, not as fixed formulas, but as conditional relationships that interact with each other.**

**5. Scenario simulations**

In our digital twin there are 3 scenarios we would want to simulate, based on what we observed in the 2022 data:

***- Scenario 1: What if 50% of nuclear capacity goes offline during a cold wave?***

The demand surges because of heating needs, while the backbone supply drops. We would have to run the gas plants to maximum which would lead to spikes in CO₂ emissions. Even the import could not be a possible option if neighboring countries face the same cold wave.

***- Scenario 2: What if solar capacity doubles while nuclear stays at current levels?***

Surplus events increase during sunny hours which means that CO₂ intensity drops on average, but with higher volatility because during night hours the system need to compensate.
So the system alternates between surplus and deficit more frequently.


***- Scenario 3: What if a million electric vehicles charge simultaneously at 7pm?***

In this case, the evening peak demand may increase by several GW.
So either imports increase or smart charging must spread the load.
